```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 5000

data = {
    "Age": np.random.randint(18, 61, n),
    "Height_cm": np.random.normal(168, 8, n).round(1),
    "Weight_kg": np.random.normal(68, 12, n).round(1),
    "Blood_Group": np.random.choice(["A+", "A-", "B+", "B-", "AB+", "AB-", "O+", "O-"], n),
    "Genotype": np.random.choice(["AA", "AS", "SS"], n),
    "Gender": np.random.choice(["Male", "Female"], n),
    "Ethnicity": np.random.choice(["Akan", "Ewe", "Ga", "Dagomba", "Other"], n),
    "Educational_Background": np.random.choice(
        ["SHS", "Diploma", "Degree", "Masters", "PhD"], n
    )
}

df = pd.DataFrame(data)

df["BMI"] = (
    df["Weight_kg"] / (df["Height_cm"] / 100) ** 2
).round(1)

df.head()
```

```python
df.to_excel("synthetic_dataset_5000.xlsx", index=False)
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["Age"], df["Height_cm"])

plt.xlabel("Age")
plt.ylabel("Height (cm)")
plt.title("Age vs Height")

plt.show()
```

```python
from sklearn.linear_model import LinearRegression

X = df[["Age"]]
y = df["Height_cm"]

model = LinearRegression()
model.fit(X, y)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
print("R²:", model.score(X, y))
```

```python
plt.scatter(df["Age"], df["Height_cm"], alpha=0.4)

plt.plot(
    df["Age"],
    model.predict(df[["Age"]])
)

plt.xlabel("Age")
plt.ylabel("Height (cm)")
plt.title("Linear Regression: Age vs Height")

plt.show()
```
